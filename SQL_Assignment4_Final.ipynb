# Assignment 4 – Data Analytics (SQL with SQLite)

Includes Questions, Queries, and Outputs.

```

import sqlite3
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

cursor.execute("CREATE TABLE employees (name TEXT, salary REAL, department TEXT, commission REAL, job_role TEXT)")
cursor.execute("CREATE TABLE students (name TEXT, marks INTEGER)")
cursor.execute("CREATE TABLE sales (region TEXT, amount REAL)")

employees_data = [
    ('A',50000,'IT',None,'Dev'),
    ('B',60000,'HR',5000,'Manager'),
    ('C',70000,'IT',None,'Dev'),
    ('D',40000,'Sales',2000,'Exec'),
    ('E',80000,'IT',None,'Lead')
]

students_data = [('S1',75),('S2',55),('S3',35),('S4',65)]
sales_data = [('North',30000),('North',25000),('South',40000),('South',20000)]

cursor.executemany("INSERT INTO employees VALUES (?,?,?,?,?)", employees_data)
cursor.executemany("INSERT INTO students VALUES (?,?)", students_data)
cursor.executemany("INSERT INTO sales VALUES (?,?)", sales_data)

conn.commit()
print("Database Ready!")
```

## Q1. Write an SQL query to display employees earning more than average salary.

```

cursor.execute("SELECT name, salary FROM employees WHERE salary > (SELECT AVG(salary) FROM employees)")
print(cursor.fetchall())
```

## Q2. Retrieve top 5 highest-paid employees.

```

cursor.execute("SELECT * FROM employees ORDER BY salary DESC LIMIT 5")
print(cursor.fetchall())
```

## Q3. Calculate total employees, average, min and max salary.

```

cursor.execute("SELECT COUNT(*), AVG(salary), MIN(salary), MAX(salary) FROM employees")
print(cursor.fetchall())
```

## Q4. Total sales per region where sales > 50000.

```

cursor.execute("SELECT region, SUM(amount) FROM sales GROUP BY region HAVING SUM(amount) > 50000")
print(cursor.fetchall())
```

## Q5. Number of unique job roles.

```

cursor.execute("SELECT COUNT(DISTINCT job_role) FROM employees")
print(cursor.fetchall())
```

## Q6. Students scoring between 60 and 80.

```

cursor.execute("SELECT * FROM students WHERE marks BETWEEN 60 AND 80")
print(cursor.fetchall())
```

## Q7. Employees with NULL commission.

```

cursor.execute("SELECT * FROM employees WHERE commission IS NULL")
print(cursor.fetchall())
```

## Q8. Increase salary of IT employees by 10%.

```

cursor.execute("UPDATE employees SET salary = salary * 1.10 WHERE department='IT'")
conn.commit()
cursor.execute("SELECT name, salary FROM employees")
print(cursor.fetchall())
```

## Q9. Delete students scoring less than 40.

```

cursor.execute("DELETE FROM students WHERE marks < 40")
conn.commit()
cursor.execute("SELECT * FROM students")
print(cursor.fetchall())
```

## Q10. Employees earning more than department average.

```

cursor.execute("""
SELECT name, salary, department
FROM employees e
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
    WHERE department = e.department
)
""")
print(cursor.fetchall())
```
